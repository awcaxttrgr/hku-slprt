# Live Handshape Detection Demo

## One-click launch (Windows)

Double-click `run_demo.bat` from File Explorer. If Conda is unavailable, it downloads and installs Miniconda for the current Windows user. On first launch, it creates the `slprt` Conda environment with Python 3.10.11 and installs `requirements.txt`; later launches reuse that environment. Internet access is required on the first run. The demo tries the webcam, then uses the bundled video if no webcam is available. Press `Q` in the display window to stop.

Run the **Preparation** code cell, then the **Running** cell. Both use the same implementation as the one-click launcher.
